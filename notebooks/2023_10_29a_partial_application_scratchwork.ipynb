# Test a function with positional arguments only

```python
@functional.partially_applicable
def f_no_kwargs2(a, b, c):
    return f'{a=}, {b=}, {c=}'
```

```python
f_no_kwargs2
```

```python
f_no_kwargs2(..., 3, 4)(5)
```

```python
f_no_kwargs2(1, 3, 4)
```

```python
f_no_kwargs2(..., 3, 4)(5)
```

```python
f_no_kwargs2(..., b=5, c=...)(3, c=7) # == 'a=3, b=5, c=7'
```

```python
f_no_kwargs2(..., b=5, c=...)(3, 7) #== 'a=3, b=5, c=7'
```

# Try an example with keyword arguments

```python
@functional.partially_applicable
def f(a, b, c, d=1, e=2):
    return f'{a=}, {b=}, {c=}, {d=}, {e=}'
```

```python
f(..., 2, ..., 4, e=...)(1, 3, 5)
```

```python
f(..., 2, ..., 4, e=...)(1, 3, e=5)
```

```python
f(..., 2, ..., 4, e=...)(1, c=3, e=5)
```

```python
f(..., 2, ..., d=..., e=...)(1, 3, e=5, d=4)
```

```python
f(..., 2, ..., d=..., e=...)(1, 3, d=4, e=5)
```

```python
f(..., 2, ..., d=..., e=...)(1, 3, e=5, x=4)
```

```python
f(..., 2, ..., e=...)(1, 3, e=5, d=12)
```

```python
f(..., 2, ..., e=...)(1, 3, e=5, d=12)
```
